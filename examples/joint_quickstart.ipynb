# JOINT quick start

d2 is the only example and reference dataset, bundled under `data/d2`.
Install JOINT from the downloaded source folder using the README instructions.
Open this notebook from the repository root or `examples` directory with that Python environment.
Outputs are written to `results/d2`. No data download or environment variable is required.

```python
from pathlib import Path

from joint import JointPipeline, load_config
```

```python
root = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "configs/d2.yaml").is_file()), None)
if root is None:
    raise FileNotFoundError("Open this notebook from the JOINT source folder or examples directory")
config = load_config(root / "configs/d2.yaml")
```

```python
pipeline = JointPipeline(config)
result = pipeline.run(resume=True, overwrite=False)
result
```

```python
from joint.pl import plot_spatial_feature

plot_spatial_feature(result, result.var_names[0])
```
